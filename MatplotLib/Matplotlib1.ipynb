```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Types of Data
#   Numerical Data
#   Categorical Data

# ak single column k upar graph plot krta hoa toa usko univariate analysis bola jata hai.
# 2 column k upar graph plot krta hoa toa usko Bivariate analysis bola jata hai.
# 2 sa jyada column k upar graph plot krta hoa toa usko multivariate analysis bola jata hai.

# 2-D line plot

# Bivariate Analysis ; 2-D line plot is used for bivariet analysis.
# categorical -> numerical and numerical -> numerical
# Use case - Time series data
```

```python
# plotting a simple function
price=[48000,54000,57000,49000,47000,45000]
year=[2015,2016,2017,2018,2019,2020]

plt.plot(year,price) #(x,y) and(categorical,numerical)
```

```python
batsman = pd.read_csv(r"D:\Coding Language\Python\DataSet\Matplotlib1\sharma-kohli.csv")
batsman
plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
# plotting multiple plots
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
# labels title
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# colors(hex) and line(width and style) and marker(size)
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='solid',linewidth=3)
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dashdot',linewidth=2)

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='solid',linewidth=3,marker='D',markersize=10)
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dashdot',linewidth=2,marker='o')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# legend -> location
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='solid',linewidth=3,marker='D',markersize=10,label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dashdot',linewidth=2,marker='o',label='Rohit')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.legend(loc='upper right') # we have to write this line to draw the line.
# By default here loc = 'best'
```

```python
price=[48000,54000,57000,49000,47000,450000] # here a very bif variation came
year=[2015,2016,2017,2018,2019,2020]

# so we see in graph for small vlaues straight line will show.

plt.plot(year,price,color="#F1A18F",linestyle='dashdot')
plt.title('price vs year')
plt.xlabel('year')
plt.ylabel('price')
plt.legend(loc='best')

# To remove that much variation we do :-
# we can do this when the graph have outlier value.
# outlier :- An outlier is a data point that is very different from the rest of the data.
plt.ylim(0,75000)
# we can also do plt.xlim(2017,2019)
```

```python
# grid
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='solid',linewidth=3,marker='D',markersize=10)
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dashdot',linewidth=2,marker='o')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.grid() # to show the grid

plt.show() # in last to show the graph
```

```python
# plt.scatter on pandas data

df = pd.read_csv(r"D:\Coding Language\Python\DataSet\Matplotlib1\batter.csv")
df = df.head(50)
plt.scatter(df['avg'],df['strike_rate'],color='red',marker='+')
plt.title('Avg and SR analysis of Top 50 Batsman')
plt.xlabel('Average')
plt.ylabel('SR')
```

```python
# size 

tips = sns.load_dataset('tips')

plt.scatter(tips['total_bill'],tips['tip'],s=tips['size']*20) 
# here s= tips['size] refers to ak sath ktna people aya hai and ushi k according
# dot ka size depend krega
# multiply by 20 kiya hai because relatively dot bht chota dkh rha tha.
```

```python
# plt.plot vs plt.scatter

# scatterplot using plt.plot 
# hum scatter type mai plot kr skta hai without using scatterplot
# then we use plt.plot and we have to put 'o'. But isma size method nhi hota hai..
# and method aisa hai joa scatter wala mai hai but plt.plot mai nhi hai.

plt.plot(tips['total_bill'],tips['tip'],'o') # This is a faster technique and scatter one is slower
```

```python
# Bar chart

"""
Bivariate Analysis
Numerical vs Categorical
Use case - Aggregate analysis of groups

"""

# Simple Bar chart

children = [10,20,40,10,30]
colors=['red','blue','green','yellow','pink']
plt.bar(colors,children,color='Green')
```

```python
# horizontal bar chart
plt.barh(colors,children,color='black') # use barh for horizontal
```

```python
# Multiple Bar charts
df = pd.read_csv(r"D:\Coding Language\Python\DataSet\Matplotlib1\batsman_season_record.csv")
print(df.shape[0])
plt.bar(np.arange(df.shape[0]) - 0.2,df['2015'],width=0.2,color='yellow')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2,color='red')
plt.bar(np.arange(df.shape[0]) + 0.2,df['2017'],width=0.2,color='blue')

plt.xticks(np.arange(df.shape[0]), df['batsman']) # To provide x-axis name.previously it was 0,1,2,4

plt.show()
```

```python
# a problem
children = [10,20,40,10,30]
colors = ['red red red red red red','blue blue blue blue','green green green green green','yellow yellow yellow yellow ','pink pinkpinkpink']

plt.bar(colors,children,color='black')
plt.xticks(rotation='vertical') # If we do not do xticks(rotation='vertical') then names will overlap.
```

```python
# Stacked Bar chart
plt.bar(df['batsman'],df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016')
plt.bar(df['batsman'],df['2015'],bottom=(df['2016'] + df['2017']),label='2015')

plt.legend()
plt.show()
```

### Pie Chart

```python
"""
Univariate/Bivariate Analysis
Categorical vs numerical
Use case - To find contibution on a standard scale
"""
# simple data
data = [23,45,100,20,49]
subjects = ['eng','science','maths','sst','hindi']
plt.pie(data,labels=subjects)

plt.show()
```

```python
# dataset
df = pd.read_csv(r"D:\Coding Language\Python\DataSet\Matplotlib1\gayle-175.csv")
df
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%') # autopct is used to show the percentage
plt.show()
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',colors=['blue','green','yellow','pink','cyan','brown'])
plt.show()
```

```python
# explode shadow
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',explode=[0.3,0,0,0,0,0.1],shadow=True) # explode is used to show the extraction
plt.show()
```
