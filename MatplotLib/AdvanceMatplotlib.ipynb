### 3D Surface plots

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
xx, yy = np.meshgrid(x,y)
```

```python
z = xx**2 + yy**2
z.shape
```

```python
fig = plt.figure(figsize=(12,8))

ax1 = fig.add_subplot(1,2,1, projection='3d')
s=ax1.plot_surface(xx, yy, z, cmap='viridis')
fig.colorbar(s)

# contour Plots :- A contour plot is a 2D representation of a 3D surface.
ax2 = fig.add_subplot(1,2,2)
p=ax2.contourf(xx,yy,z,cmap='viridis') # If we use "contour" then right side graph will not fill with colur 
fig.colorbar(p)

"""
Which approach should you use?
Situation	        Recommended
All plots are 2D	plt.subplots()
All plots are 3D	plt.subplots(subplot_kw={'projection':'3d'})
Mix of 2D and 3D	fig.add_subplot(..., projection='3d') for the 3D axis
"""

```

```python
# using "contour"
fig = plt.figure(figsize=(12,8))
z = xx**2 + yy**2
ax = plt.subplot()

p = ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
fig = plt.figure(figsize=(12,8))


z = np.sin(xx) + np.cos(yy)
# 3-D
ax1 = fig.add_subplot(1,2,1, projection='3d')
s=ax1.plot_surface(xx, yy, z, cmap='viridis')
fig.colorbar(s)

# contour Plots (2-D)
ax2 = fig.add_subplot(1,2,2)
p=ax2.contourf(xx,yy,z,cmap='viridis') 
fig.colorbar(p)
```

```python
fig = plt.figure(figsize=(12,8))


z = np.sin(xx) + np.log(xx)
# 3-D
ax1 = fig.add_subplot(1,2,1, projection='3d')
s=ax1.plot_surface(xx, yy, z, cmap='viridis')
fig.colorbar(s)

# contour Plots (2-D)
ax2 = fig.add_subplot(1,2,2)
p=ax2.contourf(xx,yy,z,cmap='viridis') 
fig.colorbar(p)
```

### Heatmap

```python
# Heatmap :- A heatmap is a 2D visualization where the color of each cell 
# represents the magnitude of the data value.
delivery = pd.read_csv(r"D:\Coding Language\Python\DataSet\datasets-session-17\deliveries.csv")
delivery.head()
print(delivery['over'].min())
```

```python
temp_df= delivery[(delivery['ball'].isin([1,2,3,4,5,6])) & (delivery['batsman_runs']==6)]
temp_df
```

```python
grid = temp_df.pivot_table(index='over',columns='ball',values='batsman_runs',aggfunc='count')
grid
```

```python
plt.figure(figsize=(20,10))
plt.imshow(grid)
plt.yticks(np.arange(20), list(range(1,21)))
plt.xticks(np.arange(0,6),list(range(1,7)))
plt.colorbar() 
```
