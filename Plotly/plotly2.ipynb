```python
match = pd.read_csv(r"D:\Coding Language\Python\DataSet\MergingJoining\matches.csv")
delivery=pd.read_csv(r"D:\Coding Language\Python\DataSet\datasets-session-17\deliveries.csv")
ipl=delivery.merge(match,left_on='match_id',right_on='id')
ipl.columns
```

```python
# Scatter Plots
# x and y both have continus data

# problem :- We are going to draw a scatter plot between Batsman Avg(X axis) and
# Batsman Strike rate(Y axis) of the top 50 batsman in IPL(ALL time)

top50=ipl.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(50).index.to_list()
new_ipl=ipl[ipl['batsman'].isin(top50)]
runs=new_ipl.groupby('batsman')['batsman_runs'].sum()
balls=new_ipl.groupby('batsman')['ball'].count()
strike=((runs/balls)*100)
strike=strike.reset_index()
strike.rename(columns={0: 'strike_rate'},inplace=True)
strike
# df=ipl[~ipl['player_dismissed'].isna()]
out=ipl['player_dismissed'].value_counts().reset_index()
runs=runs.reset_index()
strike=strike.merge(out,left_on='batsman',right_on='player_dismissed')
strike['avg']=runs['batsman_runs']/strike['count']
strike.drop(columns=['player_dismissed','count'],inplace=True)
```

```python
# Plot scatter Plot here
trace=go.Scatter(x=strike['avg'],y=strike['strike_rate'],mode='markers',text=strike['batsman'],marker={'color':'green','size':16})
data=[trace]
layout=go.Layout(title='Batsman Avg vs SR',xaxis={'title':'Batsman Average'},yaxis={'title':'Batsman strike Rate'})
fig = go.Figure(data=data,layout=layout)
pyo.plot(fig,filename='top50.html')
```

```python
#2. Line Chart :- It's an extension of Scatter plot.Usually used to show a 
# time series data.
# Year by year performance
single=ipl[ipl['batsman']=='V Kohli']
performance=single.groupby('season')['batsman_runs'].sum().reset_index()
performance

single1=ipl[ipl['batsman']=='MS Dhoni']
performance1=single1.groupby('season')['batsman_runs'].sum().reset_index()
performance1

```

```python
# Plot Line Chart here

trace=go.Scatter(x=performance['season'],y=performance['batsman_runs'],
                 mode='lines+markers',
                 marker={'color':'#00a65a'},name='Virat Kohli')

trace1=go.Scatter(x=performance1['season'],y=performance1['batsman_runs'],
                 mode='lines+markers',
                 name='MS Dhoni')

data=[trace,trace1]

layout=go.Layout(title='Year by Year Performance',
                 xaxis={'title':'Season'},
                 yaxis={'title':'Total runs'})

fig=go.Figure(data=data,layout=layout)
pyo.plot(fig)
```

```python
# Bar Plot :- Use to show relation between one categorical and 1 numerical data

# There are 3 types of Bar Graphs
# 1. Stacked Bar Graph
# 2. Stacked Bar Graph
# 3. Overlayed Bar graph
top10=ipl.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(10).reset_index()
trace=go.Bar(x=top10['batsman'],y=top10['batsman_runs'])
data=[trace]

layout=go.Layout(title='Top 10 IPL batsamn',
                 xaxis={'title':'Batsman'},
                 yaxis={'title':'Runs'})

fig=go.Figure(data=data,layout=layout)
pyo.plot(fig)
```

```python
# iw=top10.groupby(['batsman','inning'])['batsman_runs'].sum().reset_index()
list=top10.set_index('batsman').index.to_list()
top10_df=ipl[ipl['batsman'].isin(list)]
iw = top10_df.groupby(['batsman','inning'])['batsman_runs'].sum().reset_index()
one=iw[iw['inning']==1]
two=iw[iw['inning']==2]
one.rename(columns={'batsman_runs':'1st Innings'},inplace=True)
two.rename(columns={'batsman_runs':'2nd Innings'},inplace=True)
final=one.merge(two,on='batsman')[['batsman','1st Innings','2nd Innings']]
trace1 = go.Bar(x=final['batsman'],y=final['1st Innings'],name ='1st innings',
            marker={'color':'#00a65a'})

trace2 = go.Bar(x=final['batsman'],y=final['2nd Innings'],name ='2nd innings',
            marker={'color':"#0074a6"})
data=[trace1,trace2]
layout=go.Layout(title='Inning wise scores',
                 xaxis={'title':'Batsman'},
                 yaxis={'title':'runs'},
                 ) # barmode=stack/overlayed and by default nested

fig=go.Figure(data=data,layout=layout)
pyo.plot(fig)
```
