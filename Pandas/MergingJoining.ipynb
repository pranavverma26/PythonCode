```python
import numpy as np
import pandas as pd

courses=pd.read_csv(r"D:\Coding Language\Python\DataSet\MergingJoining\courses.csv")
students=pd.read_csv(r"D:\Coding Language\Python\DataSet\MergingJoining\students.csv")
nov = pd.read_csv(r"D:\Coding Language\Python\DataSet\MergingJoining\reg-month1.csv")
dec = pd.read_csv(r"D:\Coding Language\Python\DataSet\MergingJoining\reg-month2.csv")
matches=pd.read_csv(r"D:\Coding Language\Python\DataSet\MergingJoining\matches.csv")
delivery=pd.read_csv(r"D:\Coding Language\Python\DataSet\MergingJoining\deliveries (1).csv")
```

```python
# pd.concat
pd.concat([nov,dec],ignore_index=True) # If we do not have to restart the concatenate data from index
# 0 then ignore_index=False
```

```python
# differentiate both data
regs = pd.concat([nov,dec],keys=['Nov','Dec'])
regs.loc['Nov'] # from this code we got november label
# loc is applied on label index and iloc is applied on int idx
# To get the specific row of any label
regs.loc[('Nov',0)]
```

```python
# to concat across horizontal 
hori = pd.concat([nov,dec],ignore_index=True,axis=1)
hori.shape
```

```python
# In join hma doa dataframe koa join krna hota hai on the basis 
# of coumn colum.

# Inner Join:- Joins or gives only common columns
# joa phla ata hai woa left table
students.merge(regs,how='inner',on='student_id')

# In output why student_id is repeated
# The reason student_id appears multiple times is because the regs DataFrame 
# has multiple rows for the same student.


```

```python
# Left join :- here we get all value of left table and from right table
# we get common between left and right.
courses.merge(regs,how='left',on='course_id')
```

```python
# right join :- It given common data and whole right data
temp_df = pd.DataFrame({
    'student_id':[26,27,28],
    'name':['Nitish','Ankit','Rahul'],
    'partner':[28,26,17]
})
students = pd.concat([students,temp_df],ignore_index=True)
print(regs)
students.merge(regs,how='right',on='student_id')
```

```python
# outer join :- ya dono data ka column provide krta hai..
students.merge(regs,how='outer',on='student_id')
```
