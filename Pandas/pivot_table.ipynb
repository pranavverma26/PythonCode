```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
#Pivot Table :- The pivot table takes simple column-wise data as input, 
# and groups the entries into a two-dimensional table that provides a multidimensional summarization of the data.  

# Pivot table :- Pivot table generally uses on categorical columnn.

df=sns.load_dataset('tips')
df.groupby('sex')[['total_bill']].mean()
```

```python
df.groupby(['sex','smoker'])[['total_bill']].mean().unstack()

# Using pivot table we did not have to write that much thing..
df.pivot_table(index='sex',columns='smoker',values='total_bill') # here by default aggregate function is mean.
```

```python
df.pivot_table(index='sex',columns='smoker',values='total_bill',aggfunc='count') 
```

```python
# multidimensional analysis

df.pivot_table(index=['sex','smoker'],columns=['day','time'],values='total_bill',aggfunc='count')
```

```python
# This shows the power of pivot table
df.pivot_table(index=['sex','smoker'],columns=['day','time'],values=['total_bill','size','tip'],aggfunc={'total_bill':'sum','size':'mean','tip':'max'})
```

```python
# margins

df.pivot_table(index='sex',columns='smoker',values='total_bill',aggfunc='sum',margins=True) 
```
