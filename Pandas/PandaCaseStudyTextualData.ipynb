```python
# This module can correct the spelling 
from textblob import TextBlob

text = "hi I can drve at nigt"
result = TextBlob(text).correct().string
print(result)

def spelling_correction(text):
    if pd.isna(text):
        return text
    return TextBlob(text).correct().string

df['review'].apply(spelling_correction)


# Puctuations :- Aisa symbol jska koi meaning hota hai e.g ?,! etc
# Remove Punctations

import string
string.punctuation
```

```python
def remove_puctuation(text):
    
    for i in string.punctuation:
        if i in text:
            text = text.replace(i,'')
            
    return text
```

```python
df['review'] = df['review'].apply(remove_puctuation)
df['review'].head()
```

```python
# Stop word removal
from nltk.corpus import stopwords

def remove_stopwords(text):
    L=[]
    for word in text:
        if word not in stopwords.words('english'):
            L.append(word)

    return L
```

```python
df['tokenized_review']=df['tokenized_review'].apply(remove_stopwords)
df['tokenized_review']=df['tokenized_review'].apply(lambda x:" ".join(x))
```

```python
# Unigrams - contains one word
# I am Nitish My weight is 70
# Unigram-> I,am,Nitish,My,weight,is,70

# Bigrams :- contains pair of two words.
# e.g. -> [I,am],[am,Nitish],[Nitish,my]...

# trigrams :- contaons pair of three words.

from nltk import ngrams

pd.Series(ngrams(df['tokenized_review'].sum(),2)).value_counts()  # Bigrams code

pd.Series(ngrams(df['tokenized_review'].sum(),3)).value_counts()  # trigrams code

# if we find bigrams for positive and negative review both are different
```
