```python
# How to switch branch
# git checkout sidebar
# How it works :-
"""
Step 1: Your repository
Suppose your repository looks like this:
book-recommender-system/
│
├── .git/
├── app.py
└── README.md
Your working directory contains:
app.py
print("Hello")
Step 2: You make a commit
git add .
git commit -m "Initial commit"
Git stores a snapshot inside .git.
Think of .git like this:
.git/
│
├── Commit A
│     app.py -> print("Hello")
│
└── ...
Step 3: Create another branch
git checkout -b feature
Now you edit app.py:
print("Hello")
print("Feature Added")
Commit again:
git add .
git commit -m "Added feature"
Now .git contains:
.git/
Commit A
app.py
print("Hello")
↓
Commit B
app.py
print("Hello")
print("Feature Added")
Notice that both versions are stored.
Step 4: Switch back
Run:
git checkout main
Git does this internally:
Inside .git
main → Commit A
feature → Commit B
Git sees:
"User wants main."
So it reads Commit A from .git and writes those files into your working directory.
Your directory changes automatically:
Before:
print("Hello")
print("Feature Added")
After checkout:
print("Hello")
You didn't edit the file manually. Git rewrote it.
          .git
             │
             │
      ┌──────┴──────┐
      │             │
 Commit A       Commit B
 app.py         app.py
 Hello          Hello
                Feature
"""
```

```python
# Merging Branches
# What is merging ?
"""
Whenever there is merging a new commit will form.

Git merging is the process of combining changes from one branch into another. 
It allows developers to integrate work done separately.

Before Merge
main:           A ── B ── C
                      \
feature-login:         D ── E

After Merge
main:           A ── B ── C ── M
                      \       /
feature-login:         D ── E

Two types of merge :-
Fast-forward merge: Happens when the target branch has no new commits. 
Git simply moves the branch pointer forward.


Three-way merge/Regular Merge: Happens when both branches have new commits.
 Git creates a new merge commit.

"""
mergingType= cv2.imread(r"C:\Users\prana\OneDrive\Pictures\Screenshots\Screenshot 2026-07-24 013046.png")
plt.imshow(mergingType)

# To merge code :- fast-forward merging
# git checkout master
# git merge sidebar
```
