```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
import time


driver=webdriver.Chrome()
driver.get('https://www.smartprix.com/')
time.sleep(2)

link = driver.find_element(by=By.XPATH,value='//*[@id="app"]/nav[2]/ul/li[4]/a')
link.click()

# tick1=driver.find_element(by=By.XPATH,value='//*[@id="app"]/main/aside/div/div[5]/div[2]/label[1]/input')
# tick1.click()
# time.sleep(1)

# tick2=driver.find_element(by=By.XPATH,value='//*[@id="app"]/main/aside/div/div[5]/div[2]/label[2]/input')
# tick2.click()
# time.sleep(10)

old_height = driver.execute_script('return document.body.scrollHeight')
while True:
    driver.find_element(by=By.XPATH,value='//*[@id="app"]/main/div[1]/div[3]/div/div[3]').click()
    time.sleep(10)

    new_height=driver.execute_script('return document.body.scrollHeight')

    print("old ",old_height)
    print("new ",new_height)
    print("diff ",new_height-old_height)
    print()

    if new_height == old_height:
        break

    old_height=new_height


html=driver.page_source
with open(r"D:\Coding Language\Python\AdvanceWebScarapping\smartpix.html",'w',encoding='utf-8') as f:
        f.write(html)





```
