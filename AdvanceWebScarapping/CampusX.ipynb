Selenium is a suite of tools for automating web browsers. It is primarily use for testing web 

applications. But it can also be used for web scraping, data extraction, and other tools that 

require interaction with a web browser programmatically.


With Selenium, you can write code in a variety of programming languages, includding Python,Java

C#, and others, to control a web browser and interact witha web page. You can use selenium to 

simulate user interactions, such as clicking buttons, filling out forms, and navigating between

pages, to test how a web applications behaves.

In addition to its testing capabilities, selenium can also be used for web scraping, where you 

can extract information from websites and save it for further analysis or use. It is a popular 

choice for web scraping because it can interact with dynamic websites, allowing you to scraoe 

information from pages that are generated by JavaScript.

```python
# learnwith.campusx.in
# dsmp course page

# What is XPATH ?
# XPATH is a language used to navigate XML and HTML documents in order to locate elements with those documents in selenium, it is used to find elements on a web page, such as 
# links,buttons,text boxes, etc., so that they can be interacted with. XPath expressions can be used to select elements based on their tag name, attributes, or position 
# within the document:
# handling scrolling
# saving html

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time

driver=webdriver.Chrome()
driver.get("https://learnwith.campusx.in/")
time.sleep(2)

link = driver.find_element(by=By.XPATH,value='/html/body/div[1]/header/section[2]/a[1]')
link.click()
time.sleep(2)

link2=driver.find_element(by=By.XPATH,value='/html/body/div[2]/div/div[3]/div[1]/a')
link2.click()
time.sleep(2)

link3 = driver.find_element(by=By.XPATH,value='//*[@id="1698720915227d"]/div/div/div/div/div/a')
link3.click()


```
